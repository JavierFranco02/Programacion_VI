### Ejercicio 3 — Resumen de calificaciones (4 puntos)

Cree `resumen_notas(nota1, nota2, nota3)` que retorne simultáneamente la nota mayor, la nota menor y el promedio redondeado a dos decimales. Desempaquete los tres valores en variables y muéstrelos.

Pruebe con las notas `72`, `88` y `95`.

**Contenido aplicado:** retorno múltiple y funciones integradas `max()`, `min()` y `round()`.

```python
# Escriba su solución aquí
def resumen_notas(nota1, nota2, nota3):
    nota_min = min(nota1, nota2, nota3) # Obtencion de la menor nota
    nota_max = max(nota1, nota2, nota3) # Obtencion de la mayor nota 
    promedio = (nota1 + nota2 + nota3) / 3  # Calculo del promedio
    promedio = round(promedio, 2) # Redondeo del promedio a dos decimales
    return nota_min, nota_max, promedio

nota_min, nota_max, promedio = resumen_notas(72, 88, 95)
print("Nota menor:", nota_min)
print("Nota mayor:", nota_max)
print("Promedio:", promedio)
```

### Ejercicio 6 — Lista original y lista modificada (5 puntos)

Parta de `precios = [12000, 25000, 18000, 30000]`. Cree `aplicar_descuento(lista, porcentaje)` para modificar cada elemento de la lista descontando el porcentaje recibido.

1. Envíe una **copia** de `precios` a la función para aplicar un descuento del 10 %.
2. Muestre la lista modificada y compruebe que la lista original no cambió.
3. Explique brevemente qué habría ocurrido al enviar directamente la lista original.

**Contenido aplicado:** objetos mutables, paso de argumentos y copia de listas.

```python
precios = [12000, 25000, 18000, 30000]

# Escriba su solución aquí
def aplicar_descuento(lista, porcentaje):
    for i in range(len(lista)): # Iterar por cada elemento de la lista
        precio = lista[i] # Obtencion de un elemento de la lista a medida que se itera
        lista[i] = precio - (precio*porcentaje/100) # Modificacion directa del contenido de la lista para aplicarles el descuento

lista_copia = precios.copy() # Copia de la lista original de precios
aplicar_descuento(lista_copia, 10) # Llamada a la funcion para aplicar el descuento a cada precio

print("Lista de descuentos (copia modificada):", lista_copia)
print("Lista original:",precios)
```

**Explicación del estudiante:**  
Si se hubiese enviado la lista original habria sido modificado, ya que los datos complejos (como las listas) se envian por referencia, donde si la funcion modifica algun dato de la lista, esa modificacion tambien se verá reflejada en la lista original

### Ejercicio 7 — Promedio de cantidad variable de notas (5 puntos)

Cree `calcular_promedio(*notas)` para recibir cualquier cantidad de notas. Si no recibe ninguna, debe retornar un mensaje de advertencia. En caso contrario, debe retornar el promedio redondeado a dos decimales.

Pruebe la función con:

- `70, 85, 90`
- `60, 75, 80, 95, 100`
- Ningún argumento

**Contenido aplicado:** parámetros indeterminados por posición (`*args`).

```python
# Escriba su solución aquí
def calcular_promedio(*notas):
    if len(notas) != 0: # En caso de que se hayan enviado notas (donde la cantidad de elementos de la lista de notas es diferente a 0), se realiza el cálculo
        suma = 0 # Variable acumuladora donde se almacena la suma de todos los promedios
        cantidad = len(notas) # Cantidad de elementos (notas) que se han enviado
        for nota in notas: # Iteracion para sumar las notas
            suma += nota

        promedio = round(suma/cantidad, 2) # Cálculo y redondeo del promedio
        return promedio
    else:
       # Si no se enviaron notas, se devuelve una advertencia
       return "Advertencia: no se ha enviada ninguna nota para promediar"

print("Promedio 1:", calcular_promedio(70, 85, 90))
print("Promedio 2:", calcular_promedio(60, 75, 80, 95, 100))
print("Promedio 3:", calcular_promedio())
```

### Ejercicio 9 — Cuenta regresiva recursiva (6 puntos)

Cree una función recursiva `cuenta_regresiva(numero)` que muestre desde el número recibido hasta `1` y finalice mostrando `¡Inicio!`.

La función debe incluir:

- Un **caso base** que detenga las llamadas.
- Una **llamada recursiva** que reduzca el número.
- Una validación: si el valor inicial es menor que 1, debe mostrar un mensaje adecuado.

Pruebe con `5` y con `0`. Después identifique el caso base en un comentario.

**Contenido aplicado:** funciones recursivas y condición de finalización.

```python
# Escriba su solución aquí
def cuenta_regresiva(numero):
    if numero < 1:
        print("El numero debe ser mayor o igual a 1")
        return # Retornamos para no seguir ejecutando la funcion
    
    print(numero)
    if numero == 1: # Caso base, deja de llamar recursivamente cuando el numero es 1
        print("¡Inicio!")
    else:
        cuenta_regresiva(numero-1)

print("Prueba con 5:")
cuenta_regresiva(5)

print("\nPrueba con 0:")
cuenta_regresiva(0)
```

### Ejercicio 10 — Sistema de ventas escolar (7 puntos)

Desarrolle un pequeño sistema utilizando **como mínimo cuatro funciones**:

1. `calcular_subtotal(precio, cantidad)`: retorna el subtotal.
2. `calcular_descuento(subtotal, porcentaje=0)`: retorna el monto descontado.
3. `calcular_total(*subtotales)`: retorna la suma de todos los subtotales recibidos.
4. `mostrar_factura(cliente, **detalles)`: muestra el cliente y los detalles finales de la compra.

Utilice estos productos:

| Producto | Precio (Gs.) | Cantidad |
|---|---:|---:|
| Cuaderno | 18.000 | 2 |
| Regla | 5.000 | 3 |
| Mochila | 120.000 | 1 |

Aplique un descuento del 10 % al total general. La factura debe mostrar cliente, subtotal general, descuento y total final. Formatee los montos con separador de miles.

**Contenido aplicado:** integración de funciones, `return`, parámetros predeterminados, `*args`, `**kwargs` y funciones integradas.

```python
# Datos disponibles
productos = [
    {"nombre": "Cuaderno", "precio": 18000, "cantidad": 2},
    {"nombre": "Regla", "precio": 5000, "cantidad": 3},
    {"nombre": "Mochila", "precio": 120000, "cantidad": 1}
]

# Escriba su solución aquí
def calcular_subtotal(precio, cantidad):
    return cantidad*precio

# Devuelve el valor a descontar
def calcular_descuento(subtotal, porcentaje=0):
    return subtotal*porcentaje/100

def calcular_total(*subtotales):
    total = 0
    # Recorre cada subtotal y lo acumula en 'total'
    for subtotal in subtotales:
        total += subtotal
    return total

def mostrar_factura(cliente, **detalle):
    print("Cliente: ", cliente)
    # Recorre cada clave-valor de los detalles recibidos
    for clave, valor in detalle.items():
        dinero =  f"{valor:,.0f}" # Formatea el monto con separacion de miles
        dinero = dinero.replace(",", ".") # Cambia el separador de miles (antes usaba coma, ahora punto)
        print(f"{clave}: {dinero}")

subtotales = [] # Lista donde se guardan los subtotales de cada prodcuto
subtotal_general = 0

# Calcula el subtotal por cada producto
for producto in productos:
    precio = producto["precio"]
    cantidad = producto["cantidad"]
    subtotal = calcular_subtotal(precio, cantidad)
    subtotales.append(subtotal) # Almacenamos el subtotal en la lista

# Se pasa los subtotales a la funcion, se usa el operador * que desempaqueta la lista y envia cada subtotal como un argumento individual
subtotal_general = calcular_total(*subtotales)
# Calculo del descuento de 10% a partir del subtotal obtenido
monto_descuento = calcular_descuento(subtotal_general, 10)
total = subtotal_general - monto_descuento

mostrar_factura(cliente="Javier", subtotal=subtotal_general, descuento=monto_descuento, total=total)
```
